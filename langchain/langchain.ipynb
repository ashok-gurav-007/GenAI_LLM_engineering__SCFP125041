```python
from langchain_groq import ChatGroq
```

```python
import os 
```

```python
llm=ChatGroq(
    model="openai/gpt-oss-20b",
    api_key=os.environ["GROQ_API_KEY"]
)
```

```python
llm.invoke("hi")
```

```python
from langchain_core.prompts import ChatPromptTemplate
```

```python
input_temp = ChatPromptTemplate.from_messages(
    [
        ("system", "you atre experience port.write meaningful poem.do not response query other than poem"),
        ("user", "write poem about{poem_topic}")
    ]
    
)
```

```python
input_temp.invoke({"poem_topic":"rain fall"})
```

```python
response = llm.invoke("Write a meaningful poem about nature.")

print(response.content)
```

```python
from langchain_core.output_parsers import StrOutputParser

parser = StrOutputParser()
```

```python
chains = input_temp | llm | parser
```

```python
chains
```

```python
final_response = (chains.invoke({"poem_topic", "summer holidays"}))
```

```python
from langchain_core.output_parsers import JsonOutputParser
```

```python
json_parser=JsonOutputParser()
```

```python
json_parser.get_format_instructions()
```

```python
summarize_temp=ChatPromptTemplate.from_messages([
    ("system","you are expert in summarizing poem . summarize poem in 5 points {JSON_FORMATING}"),
    ("user","summarise this poem {poem}")
]).partial(JSON_FORMATING=json_parser.get_format_instructions())
```

```python
summarize_temp
```

```python
summarize_temp.invoke({"poem":final_response})
```

```python
step_2_input = summarize_temp.invoke({"poem":final_response})
```

```python
llm.invoke(step_2_input)
```

```python
step_2_resp=llm.invoke(step_2_input)
```

```python
json_parser.invoke(step_2_resp)
```

```python
chain = input_temp | llm | parser | summarize_temp | llm | json_parser
```

```python
chain.invoke({"poem_topic":"rainf fall"})
```
